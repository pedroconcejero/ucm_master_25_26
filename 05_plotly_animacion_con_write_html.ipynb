```python
import pandas as pd

import plotly.express as px
```

### Carga de datos

Los datos están incorporados a la propia librería plotly

```python
df = px.data.gapminder()
```

```python
df.describe(include = 'all')
```

## Generamos un scatterplot 

```python
px.scatter(df, 
           x="gdpPercap", 
           y="lifeExp", 
           size="pop", 
           color="continent", 
           hover_name="country",
#           log_x=True, 
           size_max=55, 
           range_x=[100,100000], 
           range_y=[25,90])
```

```python
px.scatter(df, 
           x="gdpPercap", 
           y="lifeExp", 
           animation_frame="year", 
           size="pop", 
           color="continent", 
           hover_name="country",
           log_x=True, 
           size_max=55, 
           range_x=[100,100000], 
           range_y=[25,90])
```

```python
animacion = px.scatter(df, 
           x="gdpPercap", 
           y="lifeExp", 
           animation_frame="year", 
           size="pop", 
           color="continent", 
           hover_name="country",
           log_x=True, 
           size_max=55, 
           range_x=[100,100000], 
           range_y=[25,90])
```

```python
animacion.update_layout(title_text="Relación renta per capita - esperanza de vida 1952-2007", 
                        font_size=12, 
                        height = 600)

animacion
```

```python
import plotly.io as pio
pio.write_html(animacion, 
               file='animacion.html', 
               auto_play=True)
```
